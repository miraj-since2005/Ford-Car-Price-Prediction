# EDA

```python
sns.histplot(df['price'])
plt.tight_layout()
```

```python
sns.heatmap(df.corr(numeric_only=True), annot=True)
plt.tight_layout()
```

```python
num_cols = ['year', 'mileage', 'tax', 'mpg', 'engineSize']

for i in num_cols:
    plt.figure()
    sns.scatterplot(df, x=i, y='price')
    plt.tight_layout()
```

```python
cat_cols = ['model', 'transmission', 'fuelType']

for i in cat_cols:
    plt.figure()
    sns.boxplot(df, x=i, y='price')
    plt.xticks(rotation=90)
```
